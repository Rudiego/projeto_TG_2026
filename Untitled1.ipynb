## 1. Ambiente e caminhos
Execute o Jupyter a partir da pasta do projeto. Todos os comandos usam o mesmo Python do kernel atual.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / 'config' / 'project.yaml').exists():
    raise RuntimeError('Abra o notebook com a pasta do projeto como diretório de trabalho.')

def executar(*argumentos: str) -> None:
    comando = [sys.executable, *map(str, argumentos)]
    print('Executando:', ' '.join(comando))
    subprocess.run(comando, cwd=ROOT, check=True)

print('Projeto:', ROOT)
print('Python:', sys.executable)
```

## 2. Validar imagens e bounding boxes
Coloque primeiro o dataset nas pastas indicadas no `README.md`. Imagens sem label são aceitas como exemplos de fundo; labels sem imagem e caixas inválidas interrompem o processo.

```python
executar('scripts/check_dataset.py')
```

## 3. Treinar localmente
O padrão usa YOLO26n, 640 px, batch 4 e a GTX 1050 Ti. Mude `EXECUTAR_TREINO` para `True` quando o dataset estiver validado.

```python
EXECUTAR_TREINO = True

if EXECUTAR_TREINO:
    executar('scripts/train.py')
else:
    print('Treino não iniciado. Defina EXECUTAR_TREINO = True quando estiver pronto.')
```

## 4. Avaliar o melhor checkpoint
A avaliação usa as métricas nativas de detecção do Ultralytics e gera matriz de confusão e curvas Precision-Recall/F1.

```python
EXECUTAR_AVALIACAO = False

if EXECUTAR_AVALIACAO:
    executar('scripts/evaluate.py', '--split', 'val')
else:
    print('Defina EXECUTAR_AVALIACAO = True após o treinamento.')
```
