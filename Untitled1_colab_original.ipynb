````python
from ultralytics import YOLO
import os
import pandas as pd

model_path = "/content/runs/detect/train/weights/best.pt" # Corrected path
img_dir = "/content/drive/MyDrive/dataset_2026_beta/test_2026_beta/images"
label_dir = "/content/drive/MyDrive/dataset_2026_beta/test_2026_beta/labels"

model = YOLO(model_path)

confs = [0.05,0.1,0.15,0.2,0.25,0.3,0.35,0.4,0.5]

results_table = []

for conf in confs:

    TP = 0
    FP = 0
    FN = 0

    for img in os.listdir(img_dir):
        if not img.endswith((".jpg",".png",".jpeg")):
            continue

        img_path = os.path.join(img_dir,img)
        lbl_path = os.path.join(label_dir,img.replace(".jpg",".txt").replace(".png",".txt").replace(".jpeg",".txt"))

        # ground truth
        gt = []
        if os.path.exists(lbl_path):
            with open(lbl_path) as f:
                for line in f:
                    gt.append(int(line.split()[0]))

        # predição
        pred = model(img_path,conf=conf,verbose=False)[0]

        if pred.boxes is None:
            pred_classes = []
        else:
            pred_classes = pred.boxes.cls.cpu().numpy().astype(int).tolist()

        # comparação simples
        TP += sum(1 for p in pred_classes if p in gt)
        FP += max(0,len(pred_classes)-len(gt))
        FN += max(0,len(gt)-len(pred_classes))

    precision = TP/(TP+FP+1e-6)
    recall = TP/(TP+FN+1e-6)
    f1 = 2*(precision*recall)/(precision+recall+1e-6)

    results_table.append([conf,TP,FP,FN,precision,recall,f1])

df = pd.DataFrame(results_table,
                  columns=["conf","TP","FP","FN","precision","recall","F1"])

print(df)
````

````python
from ultralytics import YOLO
import os
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc

MODEL_PATH = "/content/runs/detect/train/weights/best.pt"
VAL_IMAGES = "/content/drive/MyDrive/dataset_2026_beta/test_2026_beta/images"
VAL_LABELS = "/content/drive/MyDrive/dataset_2026_beta/test_2026_beta/labels"

model = YOLO(MODEL_PATH)

y_true = []
y_score = []

valid_ext = (".jpg",".jpeg",".png",".bmp",".webp")

for img_name in os.listdir(VAL_IMAGES):

    if not img_name.lower().endswith(valid_ext):
        continue

    img_path = os.path.join(VAL_IMAGES, img_name)
    label_path = os.path.join(VAL_LABELS, img_name.rsplit(".",1)[0] + ".txt")

    # ===== ground truth =====
    gt_has_drone = 0
    if os.path.exists(label_path):
        with open(label_path) as f:
            for line in f:
                cls = int(line.split()[0])
                if cls == 0:  # drone
                    gt_has_drone = 1
                    break

    # ===== prediction =====
    results = model(img_path, conf=0.01, verbose=False)[0]

    best_drone_conf = 0

    if results.boxes is not None:
        for box in results.boxes:
            if int(box.cls) == 0:  # drone
                conf = float(box.conf)
                best_drone_conf = max(best_drone_conf, conf)

    y_true.append(gt_has_drone)
    y_score.append(best_drone_conf)

# ===== ROC =====
fpr, tpr, thresholds = roc_curve(y_true, y_score)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(7,6))
plt.plot(fpr, tpr, linewidth=2, label=f"AUC = {roc_auc:.3f}")
plt.plot([0,1],[0,1],'--',color="gray")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Drone Detector")
plt.legend()
plt.grid()

plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
import os
from ultralytics import YOLO
from sklearn.metrics import roc_curve, auc, precision_recall_curve, average_precision_score
import cv2
from tqdm import tqdm
import glob # Import glob module

# --- CONFIGURAÇÃO ---
MODEL_PATH = "/content/runs/detect/train/weights/best.pt"
VAL_IMAGES_PATH = "/content/drive/MyDrive/dataset_2026_beta2.0/test_2026_beta/images" # Define VAL_IMAGES_PATH
CONF_THRESHOLD = 0.001  # Threshold baixo para capturar toda a curva
IOU_THRESHOLD = 0.5     # Threshold para considerar uma detecção como verdadeiro positivo

def load_ground_truth_labels(image_path):
    """
    Carrega as labels reais das imagens (assumindo formato YOLO)
    Ajuste esta função conforme seu formato de labels
    """
    label_path = image_path.replace('.jpg', '.txt').replace('.png', '.txt').replace('.jpeg', '.txt') # Added .jpeg
    label_path = label_path.replace('/images/', '/labels/')

    gt_boxes = []
    if os.path.exists(label_path):
        with open(label_path, 'r') as f:
            for line in f.readlines():
                class_id, x_center, y_center, width, height = map(float, line.strip().split())
                gt_boxes.append({
                    'class': int(class_id),
                    'bbox': [x_center, y_center, width, height]
                })
    return gt_boxes

if not os.path.exists(VAL_IMAGES_PATH):
    print(f"ERRO: A pasta '{VAL_IMAGES_PATH}' não existe!")
    print("Verifique no painel lateral esquerdo do Colab o caminho real.")
else:
    # Carregar modelo
    model = YOLO(MODEL_PATH)

    # Listar todas as imagens
    image_extensions = ['.jpg', '.jpeg', '.png', '.bmp']
    image_files = []
    for ext in image_extensions:
        image_files.extend(glob.glob(os.path.join(VAL_IMAGES_PATH, f'*{ext}')))
        image_files.extend(glob.glob(os.path.join(VAL_IMAGES_PATH, f'*{ext.upper()}')))

    print(f"Encontradas {len(image_files)} imagens para avaliação")

    # Listas para armazenar resultados
    all_scores = []
    all_labels = []

    # Processar cada imagem
    for img_path in tqdm(image_files, desc="Processando imagens"):
        # Carregar ground truth
        gt_boxes = load_ground_truth_labels(img_path)

        # Fazer predição
        results = model.predict(source=img_path, conf=CONF_THRESHOLD, verbose=False)[0]

        # Se não há ground truth, todas as detecções são falsos positivos
        if len(gt_boxes) == 0:
            if results.boxes is not None:
                for conf in results.boxes.conf.cpu().numpy():
                    all_scores.append(conf)
                    all_labels.append(0)  # False Positive
            continue

        # Se há ground truth mas nenhuma detecção
        if results.boxes is None or len(results.boxes) == 0:
            # Adicionar um placeholder para cada ground truth não detectado
            for _ in gt_boxes:
                all_scores.append(0.0)  # Score muito baixo
                all_labels.append(1)     # Era para ser detectado
            continue

        # Mapear detecções para ground truth usando IOU
        det_boxes = results.boxes.xywhn.cpu().numpy()  # normalized [x_center, y_center, width, height]
        det_confs = results.boxes.conf.cpu().numpy()
        det_classes = results.boxes.cls.cpu().numpy()

        # Matriz de IOUs
        ious = np.zeros((len(det_boxes), len(gt_boxes)))
        for i, det in enumerate(det_boxes):
            for j, gt in enumerate(gt_boxes):
                # Calcular IOU (simplificado - você pode usar uma função mais precisa)
                det_bbox = det
                gt_bbox = gt['bbox']

                # Converter para [x1, y1, x2, y2] para calcular IOU
                det_x1 = det_bbox[0] - det_bbox[2]/2
                det_y1 = det_bbox[1] - det_bbox[3]/2
                det_x2 = det_bbox[0] + det_bbox[2]/2
                det_y2 = det_bbox[1] + det_bbox[3]/2

                gt_x1 = gt_bbox[0] - gt_bbox[2]/2
                gt_y1 = gt_bbox[1] - gt_bbox[3]/2
                gt_x2 = gt_bbox[0] + gt_bbox[2]/2
                gt_y2 = gt_bbox[1] + gt_bbox[3]/2

                # Calcular interseção
                x_left = max(det_x1, gt_x1)
                y_top = max(det_y1, gt_y1)
                x_right = min(det_x2, gt_x2)
                y_bottom = min(det_y2, gt_y2)

                if x_right > x_left and y_bottom > y_top:
                    intersection = (x_right - x_left) * (y_bottom - y_top)

                    # Calcular união
                    det_area = (det_x2 - det_x1) * (det_y2 - det_y1)
                    gt_area = (gt_x2 - gt_x1) * (gt_y2 - gt_y1)
                    union = det_area + gt_area - intersection

                    ious[i, j] = intersection / union if union > 0 else 0

        # Match greedily
        matched_gt = set()
        for det_idx, conf in enumerate(det_confs):
            if len(gt_boxes) > 0:
                best_gt_idx = np.argmax(ious[det_idx])
                # Check if class matches for TP, otherwise it's FP (or ignore if classes don't match)
                # This simple mapping assumes a single class per detection for simplicity,
                # but for multi-class, more complex logic is needed
                # Assuming the GT boxes are for any class, and we are only comparing existence for simplicity
                if ious[det_idx, best_gt_idx] >= IOU_THRESHOLD and best_gt_idx not in matched_gt and det_classes[det_idx] == gt_boxes[best_gt_idx]['class']:
                    # True Positive
                    all_scores.append(conf)
                    all_labels.append(1)
                    matched_gt.add(best_gt_idx)
                else:
                    # False Positive (either low IOU, already matched GT, or class mismatch)
                    all_scores.append(conf)
                    all_labels.append(0)
            else:
                # False Positive (no ground truth in image, but detection occurred)
                all_scores.append(conf)
                all_labels.append(0)

        # Adicionar False Negatives (ground truths não detectados)
        for gt_idx in range(len(gt_boxes)):
            if gt_idx not in matched_gt:
                all_scores.append(0.0)  # Score muito baixo para FN
                all_labels.append(1)     # Era para ser detectado

    # Converter para numpy arrays
    y_true = np.array(all_labels)
    y_scores = np.array(all_scores)

    print(f"\nTotal de amostras: {len(y_true)}")
    print(f"Positivos: {np.sum(y_true)}")
    print(f"Negativos: {len(y_true) - np.sum(y_true)}")

    if len(y_true) > 0 and len(np.unique(y_true)) > 1:
        # ROC Curve
        fpr, tpr, _ = roc_curve(y_true, y_scores)
        roc_auc = auc(fpr, tpr)

        # Precision-Recall Curve
        precision, recall, _ = precision_recall_curve(y_true, y_scores)
        ap = average_precision_score(y_true, y_scores)

        # Plotting
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

        # ROC Curve
        ax1.plot(fpr, tpr, label=f'ROC (AUC = {roc_auc:.3f})', linewidth=2)
        ax1.plot([0, 1], [0, 1], 'k--', alpha=0.3)
        ax1.set_xlabel('False Positive Rate')
        ax1.set_ylabel('True Positive Rate')
        ax1.set_title('Curva ROC')
        ax1.legend()
        ax1.grid(True, alpha=0.3)

        # Precision-Recall Curve
        ax2.plot(recall, precision, label=f'PR (AP = {ap:.3f})', linewidth=2)
        ax2.set_xlabel('Recall')
        ax2.set_ylabel('Precision')
        ax2.set_title('Curva Precision-Recall')
        ax2.legend()
        ax2.grid(True, alpha=0.3)

        plt.tight_layout()
        plt.show()

        # Métricas adicionais
        print("\n--- Métricas Adicionais ---")
        print(f"AUC-ROC: {roc_auc:.4f}")
        print(f"Average Precision (AP): {ap:.4f}")

        # Calcular métricas para diferentes thresholds
        thresholds = [0.1, 0.3, 0.5, 0.7, 0.9]
        print("\n--- Métricas por Threshold ---")
        for thresh in thresholds:
            y_pred = (y_scores >= thresh).astype(int)
            tp = np.sum((y_pred == 1) & (y_true == 1))
            fp = np.sum((y_pred == 1) & (y_true == 0))
            fn = np.sum((y_pred == 0) & (y_true == 1))

            precision = tp / (tp + fp) if (tp + fp) > 0 else 0
            recall = tp / (tp + fn) if (tp + fn) > 0 else 0
            f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

            print(f"Threshold {thresh}: Precision={precision:.3f}, Recall={recall:.3f}, F1={f1:.3f}")

    else:
        print("ERRO: Não foi possível calcular as métricas. Verifique os dados.")
        print(f"Distribuição das labels: {np.unique(y_true, return_counts=True)}")
````

````python
import numpy as np
import matplotlib.pyplot as plt
import os
from ultralytics import YOLO
from sklearn.metrics import roc_curve, auc, precision_recall_curve, average_precision_score
import cv2
from tqdm import tqdm

# --- CONFIGURAÇÃO ---
MODEL_PATH = "/content/runs/detect/train/weights/best.pt"
VAL_IMAGES_PATH =
CONF_THRESHOLD = 0.001  # Threshold baixo para capturar todas as detecções
IOU_THRESHOLD = 0.5     # Threshold para considerar uma detecção como verdadeiro positivo

def load_ground_truth_labels(image_path):
    """
    Carrega as labels reais das imagens (assumindo formato YOLO)
    Ajuste esta função conforme seu formato de labels
    """
    label_path = image_path.replace('.jpg', '.txt').replace('.png', '.txt')
    label_path = label_path.replace('/images/', '/labels/')

    gt_boxes = []
    if os.path.exists(label_path):
        with open(label_path, 'r') as f:
            for line in f.readlines():
                class_id, x_center, y_center, width, height = map(float, line.strip().split())
                gt_boxes.append({
                    'class': int(class_id),
                    'bbox': [x_center, y_center, width, height]
                })
    return gt_boxes

if not os.path.exists(VAL_IMAGES_PATH):
    print(f"ERRO: A pasta '{VAL_IMAGES_PATH}' não existe!")
    print("Verifique no painel lateral esquerdo do Colab o caminho real.")
else:
    # Carregar modelo
    model = YOLO(MODEL_PATH)

    # Listar todas as imagens
    image_extensions = ['.jpg', '.jpeg', '.png', '.bmp']
    image_files = []
    for ext in image_extensions:
        image_files.extend(glob.glob(os.path.join(VAL_IMAGES_PATH, f'*{ext}')))
        image_files.extend(glob.glob(os.path.join(VAL_IMAGES_PATH, f'*{ext.upper()}')))

    print(f"Encontradas {len(image_files)} imagens para avaliação")

    # Listas para armazenar resultados
    all_scores = []
    all_labels = []

    # Processar cada imagem
    for img_path in tqdm(image_files, desc="Processando imagens"):
        # Carregar ground truth
        gt_boxes = load_ground_truth_labels(img_path)

        # Fazer predição
        results = model.predict(source=img_path, conf=CONF_THRESHOLD, verbose=False)[0]

        # Se não há ground truth, todas as detecções são falsos positivos
        if len(gt_boxes) == 0:
            if results.boxes is not None:
                for conf in results.boxes.conf.cpu().numpy():
                    all_scores.append(conf)
                    all_labels.append(0)  # False Positive
            continue

        # Se há ground truth mas nenhuma detecção
        if results.boxes is None or len(results.boxes) == 0:
            # Adicionar um placeholder para cada ground truth não detectado
            for _ in gt_boxes:
                all_scores.append(0.0)  # Score muito baixo
                all_labels.append(1)     # Era para ser detectado
            continue

        # Mapear detecções para ground truth usando IOU
        det_boxes = results.boxes.xywhn.cpu().numpy()  # normalized [x_center, y_center, width, height]
        det_confs = results.boxes.conf.cpu().numpy()
        det_classes = results.boxes.cls.cpu().numpy()

        # Matriz de IOUs
        ious = np.zeros((len(det_boxes), len(gt_boxes)))
        for i, det in enumerate(det_boxes):
            for j, gt in enumerate(gt_boxes):
                # Calcular IOU (simplificado - você pode usar uma função mais precisa)
                det_bbox = det
                gt_bbox = gt['bbox']

                # Converter para [x1, y1, x2, y2] para calcular IOU
                det_x1 = det_bbox[0] - det_bbox[2]/2
                det_y1 = det_bbox[1] - det_bbox[3]/2
                det_x2 = det_bbox[0] + det_bbox[2]/2
                det_y2 = det_bbox[1] + det_bbox[3]/2

                gt_x1 = gt_bbox[0] - gt_bbox[2]/2
                gt_y1 = gt_bbox[1] - gt_bbox[3]/2
                gt_x2 = gt_bbox[0] + gt_bbox[2]/2
                gt_y2 = gt_bbox[1] + gt_bbox[3]/2

                # Calcular interseção
                x_left = max(det_x1, gt_x1)
                y_top = max(det_y1, gt_y1)
                x_right = min(det_x2, gt_x2)
                y_bottom = min(det_y2, gt_y2)

                if x_right > x_left and y_bottom > y_top:
                    intersection = (x_right - x_left) * (y_bottom - y_top)

                    # Calcular união
                    det_area = (det_x2 - det_x1) * (det_y2 - det_y1)
                    gt_area = (gt_x2 - gt_x1) * (gt_y2 - gt_y1)
                    union = det_area + gt_area - intersection

                    ious[i, j] = intersection / union if union > 0 else 0

        # Match greedily
        matched_gt = set()
        for det_idx, conf in enumerate(det_confs):
            if len(gt_boxes) > 0:
                best_gt_idx = np.argmax(ious[det_idx])
                if ious[det_idx, best_gt_idx] >= IOU_THRESHOLD and best_gt_idx not in matched_gt:
                    # True Positive
                    all_scores.append(conf)
                    all_labels.append(1)
                    matched_gt.add(best_gt_idx)
                else:
                    # False Positive
                    all_scores.append(conf)
                    all_labels.append(0)
            else:
                # False Positive
                all_scores.append(conf)
                all_labels.append(0)

        # Adicionar False Negatives (ground truths não detectados)
        for gt_idx in range(len(gt_boxes)):
            if gt_idx not in matched_gt:
                all_scores.append(0.0)  # Score muito baixo
                all_labels.append(1)     # Era para ser detectado

    # Converter para numpy arrays
    y_true = np.array(all_labels)
    y_scores = np.array(all_scores)

    print(f"\nTotal de amostras: {len(y_true)}")
    print(f"Positivos: {np.sum(y_true)}")
    print(f"Negativos: {len(y_true) - np.sum(y_true)}")

    if len(y_true) > 0 and len(np.unique(y_true)) > 1:
        # ROC Curve
        fpr, tpr, _ = roc_curve(y_true, y_scores)
        roc_auc = auc(fpr, tpr)

        # Precision-Recall Curve
        precision, recall, _ = precision_recall_curve(y_true, y_scores)
        ap = average_precision_score(y_true, y_scores)

        # Plotting
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

        # ROC Curve
        ax1.plot(fpr, tpr, label=f'ROC (AUC = {roc_auc:.3f})', linewidth=2)
        ax1.plot([0, 1], [0, 1], 'k--', alpha=0.3)
        ax1.set_xlabel('False Positive Rate')
        ax1.set_ylabel('True Positive Rate')
        ax1.set_title('Curva ROC')
        ax1.legend()
        ax1.grid(True, alpha=0.3)

        # Precision-Recall Curve
        ax2.plot(recall, precision, label=f'PR (AP = {ap:.3f})', linewidth=2)
        ax2.set_xlabel('Recall')
        ax2.set_ylabel('Precision')
        ax2.set_title('Curva Precision-Recall')
        ax2.legend()
        ax2.grid(True, alpha=0.3)

        plt.tight_layout()
        plt.show()

        # Métricas adicionais
        print("\n--- Métricas Adicionais ---")
        print(f"AUC-ROC: {roc_auc:.4f}")
        print(f"Average Precision (AP): {ap:.4f}")

        # Calcular métricas para diferentes thresholds
        thresholds = [0.1, 0.3, 0.5, 0.7, 0.9]
        print("\n--- Métricas por Threshold ---")
        for thresh in thresholds:
            y_pred = (y_scores >= thresh).astype(int)
            tp = np.sum((y_pred == 1) & (y_true == 1))
            fp = np.sum((y_pred == 1) & (y_true == 0))
            fn = np.sum((y_pred == 0) & (y_true == 1))

            precision = tp / (tp + fp) if (tp + fp) > 0 else 0
            recall = tp / (tp + fn) if (tp + fn) > 0 else 0
            f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

            print(f"Threshold {thresh}: Precision={precision:.3f}, Recall={recall:.3f}, F1={f1:.3f}")

    else:
        print("ERRO: Não foi possível calcular as métricas. Verifique os dados.")
        print(f"Distribuição das labels: {np.unique(y_true, return_counts=True)}")
````
